Example 1: Given a string s, return true if it is a palindrome, false otherwise.

A string is a palindrome if it reads the same forward as backward. That means, after reversing it, it is still the same string. For example: "abcdcba", or "racecar".

```python
def is_palindrome(arr: List[Any]):
    left = 0
    right = len(arr)-1
    
    while left < right:
        if arr[left] == arr[right]:
            left+=1
            right-=1
        else:
             return f"{arr} is not a palindrome"
    
    return f"{arr} is a palindrome"
        
    
```

```python
is_palindrome(arr='ben')
```

```python
def check_if_palindrome(s):
    left = 0
    right = len(s) - 1

    while left < right:
        if s[left] != s[right]:
            return False
        left += 1
        right -= 1
    
    return True
```

A subsequence of a string is a sequence of characters that can be obtained by deleting some (or none) of the characters from the original string, while maintaining the relative order of the remaining characters. For example, "ace" is a subsequence of "abcde" while "aec" is not.

```python
def is_subsequence(str_1:str, str_2:str)->bool:
    i = j = 0
    
    if len(str_1) > len(str_2):
        return False

    while i < len(str_1) and j < len(str_2):
        print(str_1[i], str_2[j])
        if str_1[i] == str_2[j]:
            i+=1
        j+=1

    return i == len(str_1)
            
    #return False
```

You must do this by modifying the input array in-place with O(1) extra memory.

```python
class Solution:
    def reverseString(self, s: List[str]) -> None:
        """
        Do not return anything, modify s in-place instead.
        """
        s[:] = s[::-1]

solution = Solution()
s=["h","e","l","l","o"]

# --------------------------------------------------|

solution.reverseString(s)
print(s)
```

```python
class Solution:
    def reverseString(self, s: List[str]) -> None:
        left, right = 0, len(s) - 1
        
        while left < right:
            # Swap elements at left and right indices
            s[left], s[right] = s[right], s[left]
            
            # Move pointers inward
            left += 1
            right -= 1
```

```python
s=["h","e","l","l","o"]
solution = Solution()
solution.reverseString(s)
s
```

Example 1: Given an array of positive integers nums and an integer k, find the length of the longest subarray whose sum is less than or equal to k.

```python
nums = [3, 1, 2, 7, 4, 2, 1, 1, 5]
k = 8
```

```python
def find_length(nums, k):
    # curr is the current sum of the window
    left = curr = ans = 0
    for right in range(len(nums)):
        curr += nums[right]
        print(curr)
        while curr > k:
            curr -= nums[left]
            left += 1
            
        ans = max(ans, right - left + 1)
    
    return ans
```

```python
find_length(nums, k)
```

Given an array of integers nums and an integer target, return indices of two numbers such that they add up to target. You cannot use the same index twice.

```python
class Solution:
    def twoSum(self, nums:List[int], target:int):
        dict = {}
        for i in range(len(nums)):
            num = nums[i]
            complement = target - num
            if complement in dict:
                return [i, dict[complement]]
            dict[num] = i
        return [-1,-1]
```

Example 2: 2351. First Letter to Appear Twice

Given a string s, return the first character to appear twice. It is guaranteed that the input will have a duplicate character.

```python
class Solution:
    def duplicate_char(self, s:str):
        dic = {}
        for i in range(len(s)):
            if s[i] in dic.values():
                return s[i]
            dic[i] = s[i] 
        return "No duplicate charaters"
```

```python
class Solution:
    def duplicate_char(self, s:str):
        seen = set()
        for c in s:
            if c in seen:
                return c
            seen.add(c) 
            
        return "No duplicate charaters"
```
